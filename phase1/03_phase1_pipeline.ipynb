## Continuing decoder training

The overnight run disconnected after 6.6 hours (Colab's free-tier session limit) at step 500/3100, with validation loss already down to 3.1. Since the original target of 3100 steps would take too long on a free GPU, lowering the token budget here to a more realistic total and resuming training from the saved checkpoint instead of starting over.


```python
!python train.py --arch decoder \
    --data_dir /content/drive/MyDrive/CS781/data_p1 \
    --out_dir /content/drive/MyDrive/CS781/ckpt_p1/decoder \
    --total_tokens 1.8e8
```
